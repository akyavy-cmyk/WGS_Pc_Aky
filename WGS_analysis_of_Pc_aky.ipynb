```python
"""
Complete genome analysis for an NCBI RefSeq assembly (Google Colab ready)
=========================================================================
Inputs (upload all four when prompted; .gz files are accepted too):
    *_assembly_report.txt   *_genomic.fna   *_genomic.gff   *_protein.faa

Analyses
    1. Assembly overview      : replicons, size, GC, N50, ambiguous bases
    2. Sequence composition   : windowed GC, GC skew, cumulative skew -> oriC/terminus
                                estimate, dinucleotide signature
    3. Gene structure (GFF)   : gene classes, lengths, genic fraction, intergenic gaps,
                                overlaps, strand bias, leading/lagging strand
    4. RNA genes & pseudogenes: rRNA operons, tRNA inventory, pseudogene table,
                                shared protein IDs
    5. Codon usage            : GC3s, RSCU, ENC (Wright plot), CAI, atypical genes,
                                tRNA supply vs codon demand
    6. Proteins (FAA)         : length, MW, pI, GRAVY, amino-acid composition,
                                hypothetical fraction
    7. Circular genome map(s)

HOW TO USE IN COLAB
    1. https://colab.research.google.com -> New notebook
    2. Paste this whole script in ONE cell and run it
    3. Click "Choose files" and select all four files at once
    4. Plots appear inline; results.zip (tables + figures) downloads at the end

Only needs pandas, numpy and matplotlib (all preinstalled in Colab).
Outside Colab:  python genome_analysis_colab.py report.txt genome.fna genes.gff proteins.faa
"""
import gzip
import itertools
import math
import re
import shutil
import sys
import warnings
from collections import Counter, defaultdict
from pathlib import Path
from urllib.parse import unquote

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

warnings.filterwarnings("ignore")
plt.rcParams.update({"axes.spines.top": False, "axes.spines.right": False})

# ----------------------------------------------------------------------
# Settings you may want to change
# ----------------------------------------------------------------------
OUT = Path("results")            # output folder
GC_WINDOW = None                 # window size in bp for GC / skew (None = automatic)
MIN_REPLICON_FOR_MAP = 50_000    # smallest replicon that gets a circular map
MAX_MAPS = 4                     # maximum number of circular maps
MIN_CODONS_PER_GENE = 80         # genes shorter than this are skipped for ENC / CAI

SUMMARY = []                     # lines of the final text summary


# ----------------------------------------------------------------------
# Generic helpers
# ----------------------------------------------------------------------
def say(text=""):
    print(text)
    SUMMARY.append(text)


def heading(title):
    say("\n" + "=" * 70)
    say(title)
    say("=" * 70)


def open_text(path):
    p = str(path)
    return gzip.open(p, "rt") if p.endswith(".gz") else open(p, "r")


def save_fig(fig, name):
    fig.tight_layout()
    fig.savefig(OUT / name, dpi=200, bbox_inches="tight")
    plt.show()
    plt.close(fig)


def save_table(df, name, index=False):
    df.to_csv(OUT / name, index=index)


def show(df, n=15):
    print(df.head(n).to_string(index=False))
    if len(df) > n:
        print(f"... ({len(df)} rows in total, full table saved to CSV)")


def read_fasta(path):
    seqs, headers, name, buf = {}, {}, None, []
    with open_text(path) as f:
        for line in f:
            line = line.rstrip()
            if line.startswith(">"):
                if name is not None:
                    seqs[name] = "".join(buf)
                headers[line[1:].split()[0]] = line[1:]
                name = line[1:].split()[0]
                buf = []
            elif name is not None:
                buf.append(line.strip())
    if name is not None:
        seqs[name] = "".join(buf)
    return seqs, headers


COMP = str.maketrans("ACGTNacgtn", "TGCANtgcan")


def revcomp(s):
    return s.translate(COMP)[::-1]


def gc_frac(s):
    g, c = s.count("G") + s.count("C"), s.count("A") + s.count("T")
    return g / (g + c) if (g + c) else float("nan")


def n50(lengths):
    lengths = sorted(lengths, reverse=True)
    half, acc = sum(lengths) / 2, 0
    for x in lengths:
        acc += x
        if acc >= half:
            return x
    return 0


def merged_length(starts, ends):
    iv = sorted(zip(starts, ends))
    total, cur_s, cur_e = 0, None, None
    for s, e in iv:
        if cur_e is None or s > cur_e + 1:
            if cur_e is not None:
                total += cur_e - cur_s + 1
            cur_s, cur_e = s, e
        else:
            cur_e = max(cur_e, e)
    if cur_e is not None:
        total += cur_e - cur_s + 1
    return total


# ----------------------------------------------------------------------
# Input handling
# ----------------------------------------------------------------------
def classify_files(names):
    found = {}
    for n in names:
        low = n.lower()
        if "assembly_report" in low:
            found["report"] = n
        elif re.search(r"\.(fna|fa|fasta|fas)(\.gz)?$", low):
            if "cds_from" in low or "rna_from" in low:
                print(f"WARNING: {n} looks like a CDS/RNA subset, not the genomic FASTA; ignored.")
                continue
            found["fna"] = n
        elif re.search(r"\.(gff|gff3)(\.gz)?$", low):
            found["gff"] = n
        elif re.search(r"\.faa(\.gz)?$", low):
            found["faa"] = n
    return found


def get_input_files():
    try:
        from google.colab import files  # only available in Colab
        print("Select all four files (assembly report, .fna, .gff, .faa) and upload them...")
        uploaded = files.upload()
        names = list(uploaded)
    except ImportError:
        names = sys.argv[1:]
        if not names:
            raise SystemExit("Usage: python genome_analysis_colab.py report.txt genome.fna genes.gff proteins.faa")
    found = classify_files(names)
    for need in ("fna", "gff"):
        if need not in found:
            raise SystemExit(f"Missing required file: {need} (the .{need} file). Please upload it and run again.")
    for opt, label in (("report", "assembly report"), ("faa", "protein .faa")):
        if opt not in found:
            print(f"Note: no {label} found - the related analysis will be skipped.")
    print("Using:", found)
    return found


# ----------------------------------------------------------------------
# 1. Assembly report and genome overview
# ----------------------------------------------------------------------
def parse_assembly_report(path):
    meta, header, rows = {}, None, []
    with open_text(path) as f:
        for line in f:
            line = line.rstrip("\r\n")
            if line.startswith("# Sequence-Name"):
                header = line[1:].strip().split("\t")
            elif line.startswith("#"):
                m = re.match(r"#\s*([^:]+):\s*(.*)$", line)
                if m:
                    meta[m.group(1).strip()] = m.group(2).strip()
            elif line.strip():
                rows.append(line.split("\t"))
    tbl = pd.DataFrame(rows)
    if header and not tbl.empty and tbl.shape[1] == len(header):
        tbl.columns = header
    return meta, tbl


def replicon_table(genome, report_tbl):
    info = {}
    if report_tbl is not None and not report_tbl.empty:
        for _, r in report_tbl.iterrows():
            for col in ("RefSeq-Accn", "GenBank-Accn", "Sequence-Name"):
                if col in report_tbl.columns:
                    info[str(r[col])] = r
    rows = []
    for sid, seq in genome.items():
        acgt = sum(seq.count(b) for b in "ACGT")
        r = info.get(sid)
        rows.append({
            "seqid": sid,
            "name": r.get("Sequence-Name", "") if r is not None else "",
            "type": r.get("Assigned-Molecule-Location/Type", "") if r is not None else "",
            "length_bp": len(seq),
            "GC_percent": round(gc_frac(seq) * 100, 2),
            "N_count": seq.count("N"),
            "other_ambiguous": len(seq) - acgt - seq.count("N"),
        })
    return pd.DataFrame(rows).sort_values("length_bp", ascending=False).reset_index(drop=True)


def plot_overview(rep):
    d = rep.head(20)
    fig, ax = plt.subplots(1, 2, figsize=(12, 4))
    ax[0].bar(d["seqid"], d["length_bp"] / 1e6, color="steelblue")
    ax[0].set_ylabel("Length (Mb)")
    ax[0].set_title("Replicon / contig sizes")
    ax[1].bar(d["seqid"], d["GC_percent"], color="seagreen")
    ax[1].set_ylabel("GC (%)")
    ax[1].set_title("GC content per replicon")
    ax[1].set_ylim(max(0, d["GC_percent"].min() - 10), min(100, d["GC_percent"].max() + 10))
    for a in ax:
        a.tick_params(axis="x", rotation=60)
    save_fig(fig, "01_replicon_overview.png")


# ----------------------------------------------------------------------
# 2. Sequence composition: windows, skew, origin, dinucleotides
# ----------------------------------------------------------------------
def auto_window(L):
    if GC_WINDOW:
        return min(GC_WINDOW, max(L // 4, 1))
    return int(max(500, round(L / 600 / 100) * 100))


def window_stats(seq, win):
    a = np.frombuffer(seq.encode(), dtype=np.uint8)
    cs = lambda m: np.concatenate([[0], np.cumsum(m)])
    G, C, A, T = cs(a == 71), cs(a == 67), cs(a == 65), cs(a == 84)
    starts = np.arange(0, len(a) - win + 1, win)
    ends = starts + win
    g, c, at = G[ends] - G[starts], C[ends] - C[starts], (A[ends] - A[starts]) + (T[ends] - T[starts])
    with np.errstate(divide="ignore", invalid="ignore"):
        gc = (g + c) / (g + c + at)
        skew = (g - c) / (g + c)
    return starts + win // 2, gc * 100, skew


def find_ori_ter(pos, skew):
    cum = np.cumsum(np.nan_to_num(skew))
    return int(pos[np.argmin(cum)]), int(pos[np.argmax(cum)]), cum


def dinucleotide_rho(seq):
    lut = np.full(256, -1, dtype=np.int64)
    for i, b in enumerate("ACGT"):
        lut[ord(b)] = i
    a = lut[np.frombuffer(seq.encode(), dtype=np.uint8)]
    comp = np.where(a >= 0, 3 - a, -1)[::-1]
    mat = np.zeros(16)
    for arr in (a, comp):
        x, y = arr[:-1], arr[1:]
        ok = (x >= 0) & (y >= 0)
        mat += np.bincount(x[ok] * 4 + y[ok], minlength=16)
    mat = mat.reshape(4, 4) / mat.sum()
    f = mat.sum(axis=1)
    return pd.DataFrame(mat / np.outer(f, f), index=list("ACGT"), columns=list("ACGT"))


def composition_analysis(genome, rep):
    heading("2. SEQUENCE COMPOSITION")
    total = rep["length_bp"].sum()
    main = rep.iloc[0]
    complete_like = main["length_bp"] >= 0.5 * total
    big = rep[rep["length_bp"] >= 100_000].head(4)
    if big.empty:
        big = rep.head(1)
    ori_ter, win_data = {}, {}

    fig, axes = plt.subplots(3, len(big), figsize=(6 * len(big), 8), squeeze=False)
    for j, (_, r) in enumerate(big.iterrows()):
        sid, seq = r["seqid"], genome[r["seqid"]]
        win = auto_window(len(seq))
        pos, gc, skew = window_stats(seq, win)
        win_data[sid] = (pos, gc, skew)
        ori, ter, cum = find_ori_ter(pos, skew)
        if sid == main["seqid"] and complete_like and len(pos) >= 20:
            ori_ter[sid] = (ori, ter)
        x = pos / 1e6
        axes[0, j].plot(x, gc, lw=0.7, color="seagreen")
        axes[0, j].axhline(np.nanmean(gc), color="grey", ls="--", lw=0.6)
        axes[0, j].set_title(f"{sid}\nGC % ({win} bp windows)")
        axes[1, j].fill_between(x, 0, np.nan_to_num(skew), where=np.nan_to_num(skew) >= 0, color="#1b9e77")
        axes[1, j].fill_between(x, 0, np.nan_to_num(skew), where=np.nan_to_num(skew) < 0, color="#7570b3")
        axes[1, j].set_title("GC skew (G-C)/(G+C)")
        axes[2, j].plot(x, cum, color="black", lw=0.8)
        axes[2, j].set_title("Cumulative GC skew")
        axes[2, j].set_xlabel("Position (Mb)")
        if sid in ori_ter:
            for k in range(3):
                axes[k, j].axvline(ori / 1e6, color="red", ls="--", lw=0.8)
                axes[k, j].axvline(ter / 1e6, color="blue", ls="--", lw=0.8)
            axes[2, j].text(ori / 1e6, cum.min(), " oriC", color="red", fontsize=8)
            axes[2, j].text(ter / 1e6, cum.max(), " ter", color="blue", fontsize=8)
    save_fig(fig, "02_gc_and_skew.png")

    if ori_ter:
        sid = main["seqid"]
        o, t = ori_ter[sid]
        say(f"Estimated origin (min cumulative skew) of {sid}: ~{o:,} bp; terminus (max): ~{t:,} bp")
        say("  (skew-based estimate; verify against dnaA location / literature)")
    else:
        say("Origin estimate skipped (largest replicon is <50% of the assembly, i.e. probably a draft).")

    # Dinucleotide signature of the main replicon
    rho = dinucleotide_rho(genome[main["seqid"]])
    save_table(rho, "02_dinucleotide_rho.csv", index=True)
    fig, ax = plt.subplots(figsize=(4.5, 4))
    im = ax.imshow(rho.values, cmap="coolwarm", vmin=0.6, vmax=1.4)
    ax.set_xticks(range(4), list(rho.columns))
    ax.set_yticks(range(4), list(rho.index))
    for i in range(4):
        for j in range(4):
            ax.text(j, i, f"{rho.values[i, j]:.2f}", ha="center", va="center", fontsize=8)
    ax.set_title("Dinucleotide relative abundance (rho)")
    fig.colorbar(im, ax=ax, shrink=0.8)
    save_fig(fig, "02_dinucleotide_rho.png")
    return ori_ter, win_data


# ----------------------------------------------------------------------
# GFF parsing
# ----------------------------------------------------------------------
GFF_KEYS = ["ID", "Parent", "Name", "locus_tag", "gene", "gene_biotype", "product", "protein_id",
            "pseudo", "Note", "note", "anticodon", "transl_table", "partial", "gbkey"]


def parse_attrs(s):
    d = {}
    for part in s.strip().split(";"):
        if "=" in part:
            k, v = part.split("=", 1)
            d[k] = unquote(v)
    return d


def parse_gff(path):
    recs = []
    with open_text(path) as f:
        for line in f:
            if line.startswith("#") or not line.strip():
                continue
            p = line.rstrip("\r\n").split("\t")
            if len(p) < 9:
                continue
            try:
                s, e = int(p[3]), int(p[4])
            except ValueError:
                continue
            at = parse_attrs(p[8])
            rec = {"seqid": p[0], "type": p[2], "start": s, "end": e, "strand": p[6], "phase": p[7]}
            rec.update({k: at.get(k, "") for k in GFF_KEYS})
            recs.append(rec)
    df = pd.DataFrame(recs)
    df["length"] = df["end"] - df["start"] + 1
    df["note_all"] = df["Note"].where(df["Note"] != "", df["note"])
    return df


def build_gene_table(gff):
    g = gff[gff["type"].isin(["gene", "pseudogene"])].copy()
    child = gff[gff["Parent"] != ""]
    first_child = child.drop_duplicates("Parent").set_index("Parent")
    notes = child[child["note_all"] != ""].drop_duplicates("Parent").set_index("Parent")["note_all"]
    g["child_type"] = g["ID"].map(first_child["type"]).fillna("")
    g["product"] = g["ID"].map(first_child["product"]).fillna("")
    g["note_all"] = g["note_all"].where(g["note_all"] != "", g["ID"].map(notes).fillna(""))
    g["pseudo_flag"] = (
        (g["type"] == "pseudogene")
        | (g["pseudo"].str.lower() == "true")
        | g["gene_biotype"].str.contains("pseudo", case=False)
    )
    g["class"] = np.where(g["pseudo_flag"], "pseudogene",
                          g["gene_biotype"].where(g["gene_biotype"] != "", g["child_type"].replace("", "other")))
    g["mid"] = (g["start"] + g["end"]) // 2
    return g.reset_index(drop=True)


# ----------------------------------------------------------------------
# 3. Gene structure
# ----------------------------------------------------------------------
def is_leading(mid, strand, ori, ter, L):
    first_arc = ((mid - ori) % L) <= ((ter - ori) % L)
    return (strand == "+") == first_arc


def gene_structure(genes, gff, rep, ori_ter):
    heading("3. GENE STRUCTURE (GFF)")
    comp = genes["class"].value_counts().rename_axis("gene_class").reset_index(name="count")
    comp["percent"] = (comp["count"] / comp["count"].sum() * 100).round(2)
    save_table(comp, "03_gene_classes.csv")
    say(f"Total genes (incl. pseudogenes): {len(genes):,}")
    show(comp)

    lens = {r["seqid"]: r["length_bp"] for _, r in rep.iterrows()}
    rows = []
    for sid, d in genes.groupby("seqid"):
        L = lens.get(sid, np.nan)
        cds = gff[(gff["seqid"] == sid) & (gff["type"] == "CDS")]
        rows.append({
            "seqid": sid, "length_bp": L, "genes": len(d),
            "genes_per_Mb": round(len(d) / L * 1e6, 1) if L else np.nan,
            "genic_fraction_%": round(merged_length(d["start"], d["end"]) / L * 100, 2) if L else np.nan,
            "CDS_fraction_%": round(merged_length(cds["start"], cds["end"]) / L * 100, 2) if L and len(cds) else 0,
            "plus_strand": int((d["strand"] == "+").sum()),
            "minus_strand": int((d["strand"] == "-").sum()),
        })
    per_rep = pd.DataFrame(rows).sort_values("length_bp", ascending=False)
    save_table(per_rep, "03_per_replicon_gene_stats.csv")
    show(per_rep)

    # Intergenic gaps and overlaps
    gaps = []
    for sid, d in genes.sort_values(["seqid", "start"]).groupby("seqid"):
        s, e, st = d["start"].values, d["end"].values, d["strand"].values
        if len(s) < 2:
            continue
        prev_end = np.maximum.accumulate(e)[:-1]
        gaps.append(pd.DataFrame({"seqid": sid, "gap": s[1:] - prev_end - 1, "same_strand": st[1:] == st[:-1]}))
    gaps = pd.concat(gaps) if gaps else pd.DataFrame(columns=["seqid", "gap", "same_strand"])
    overl = gaps[gaps["gap"] < 0]["gap"].value_counts().head(6)
    say(f"\nOverlapping neighbouring genes: {int((gaps['gap'] < 0).sum()):,} "
        f"({(gaps['gap'] < 0).mean() * 100:.1f}%). Most common overlaps (bp): "
        + ", ".join(f"{-k} bp x{v}" for k, v in overl.items()))
    say(f"Median intergenic gap: {gaps['gap'].median():.0f} bp; "
        f"adjacent genes on the same strand: {gaps['same_strand'].mean() * 100:.1f}%")

    # Leading / lagging strand
    lead_info = None
    main_id = rep.iloc[0]["seqid"]
    if main_id in ori_ter:
        ori, ter = ori_ter[main_id]
        d = genes[(genes["seqid"] == main_id) & (genes["class"] != "pseudogene")].copy()
        d["leading"] = [is_leading(m, s, ori, ter, lens[main_id]) for m, s in zip(d["mid"], d["strand"])]
        lead_info = d
        pc = d[d["class"] == "protein_coding"]
        say(f"Protein-coding genes on the leading strand: {pc['leading'].mean() * 100:.1f}% "
            f"({int(pc['leading'].sum()):,} of {len(pc):,})")
        for cls in ("rRNA", "tRNA"):
            x = d[d["class"] == cls]
            if len(x):
                say(f"{cls} genes on the leading strand: {x['leading'].mean() * 100:.1f}% ({int(x['leading'].sum())} of {len(x)})")

    # Plots
    fig, ax = plt.subplots(1, 4, figsize=(18, 4))
    pc = genes[genes["class"] == "protein_coding"]
    ax[0].hist(pc["length"], bins=60, color="steelblue")
    ax[0].axvline(pc["length"].median(), color="red", ls="--")
    ax[0].set_title(f"Protein-coding gene length (median {pc['length'].median():.0f} bp)")
    ax[0].set_xlabel("bp")
    ax[1].hist(gaps["gap"].clip(-60, 600), bins=70, color="darkorange")
    ax[1].set_title("Intergenic distance (negative = overlap)")
    ax[1].set_xlabel("bp (clipped to -60..600)")
    top = comp.head(8)
    ax[2].barh(top["gene_class"][::-1], top["count"][::-1], color="seagreen")
    ax[2].set_xscale("log")
    ax[2].set_title("Gene classes (log scale)")
    if lead_info is not None:
        cats = ["protein_coding", "rRNA", "tRNA", "pseudogene"]
        allg = genes[genes["seqid"] == main_id].copy()
        allg["leading"] = [is_leading(m, s, *ori_ter[main_id], lens[main_id]) for m, s in zip(allg["mid"], allg["strand"])]
        pct = [allg[allg["class"] == c]["leading"].mean() * 100 if (allg["class"] == c).any() else np.nan for c in cats]
        ax[3].bar(cats, pct, color="slateblue")
        ax[3].axhline(50, color="grey", ls="--")
        ax[3].set_ylabel("% on leading strand")
        ax[3].set_title("Leading-strand bias")
        ax[3].tick_params(axis="x", rotation=30)
    else:
        ax[3].bar(per_rep["seqid"].head(8), per_rep["plus_strand"].head(8), label="+")
        ax[3].bar(per_rep["seqid"].head(8), per_rep["minus_strand"].head(8),
                  bottom=per_rep["plus_strand"].head(8), label="-")
        ax[3].legend()
        ax[3].set_title("Genes per strand")
        ax[3].tick_params(axis="x", rotation=60)
    save_fig(fig, "03_gene_structure.png")
    return comp, gaps


# ----------------------------------------------------------------------
# 4. RNA genes, pseudogenes, shared protein ids
# ----------------------------------------------------------------------
THREE_TO_ONE = {"Ala": "A", "Arg": "R", "Asn": "N", "Asp": "D", "Cys": "C", "Gln": "Q", "Glu": "E",
                "Gly": "G", "His": "H", "Ile": "I", "Leu": "L", "Lys": "K", "Met": "M", "fMet": "M",
                "iMet": "M", "Phe": "F", "Pro": "P", "Ser": "S", "Thr": "T", "Trp": "W", "Tyr": "Y",
                "Val": "V", "Sec": "U"}


def rna_analysis(gff, genes, rep, ori_ter):
    heading("4. RNA GENES, PSEUDOGENES AND SHARED PROTEIN IDS")
    lens = dict(zip(rep["seqid"], rep["length_bp"]))
    main_id = rep.iloc[0]["seqid"]

    # ---- rRNA operons
    r = gff[gff["type"] == "rRNA"].sort_values(["seqid", "start"]).copy()

    def rtype(p):
        m = re.search(r"\b(5S|16S|23S)\b", p)
        return m.group(1) if m else "other"

    r["rrna_type"] = r["product"].map(rtype)
    ops = []
    for sid, d in r.groupby("seqid"):
        cluster = []
        for _, x in d.iterrows():
            if cluster and x["start"] - cluster[-1]["end"] > 5000:
                ops.append((sid, cluster))
                cluster = []
            cluster.append(x)
        if cluster:
            ops.append((sid, cluster))
    op_rows = []
    for sid, c in ops:
        mid = (c[0]["start"] + c[-1]["end"]) // 2
        rel = np.nan
        if sid == main_id and main_id in ori_ter:
            L, ori = lens[sid], ori_ter[sid][0]
            dist = min((mid - ori) % L, (ori - mid) % L)
            rel = round(dist / (L / 2), 2)   # 0 = at oriC, 1 = at terminus
        op_rows.append({"seqid": sid, "start": c[0]["start"], "end": c[-1]["end"], "strand": c[0]["strand"],
                        "composition": "-".join(x["rrna_type"] for x in c), "n_genes": len(c),
                        "relative_distance_from_oriC": rel})
    ops_df = pd.DataFrame(op_rows)
    save_table(ops_df, "04_rrna_operons.csv")
    cnt = r["rrna_type"].value_counts().to_dict()
    say(f"rRNA genes: {len(r)} ({cnt}); grouped into {len(ops_df)} operon(s)/clusters")
    if len(ops_df):
        show(ops_df, 10)
        if ops_df["relative_distance_from_oriC"].notna().any():
            say(f"Mean relative distance of rRNA clusters from oriC (0=ori, 1=ter): "
                f"{ops_df['relative_distance_from_oriC'].mean():.2f}")

    # ---- tRNA inventory
    t = gff[gff["type"] == "tRNA"].copy()

    def anticodon(a):
        m = re.search(r"seq:([A-Za-z]+)", a)
        return m.group(1).upper() if m else ""

    def aa_name(row):
        m = re.search(r"aa:([A-Za-z]+)", row["anticodon"])
        if m:
            return m.group(1)
        return re.sub(r"^tRNA-", "", row["product"]).split("(")[0]

    t["aa3"] = t.apply(aa_name, axis=1)
    t["anticodon_seq"] = t["anticodon"].map(anticodon)
    t["aa1"] = t["aa3"].map(THREE_TO_ONE)
    save_table(t[["seqid", "start", "end", "strand", "product", "aa3", "anticodon_seq"]], "04_trna_table.csv")
    say(f"\ntRNA genes: {len(t)}; distinct amino acids covered: {t['aa1'].nunique()} "
        f"(of 20); distinct anticodons: {t['anticodon_seq'].replace('', np.nan).nunique()}")

    # ---- pseudogenes
    p = genes[genes["class"] == "pseudogene"].copy()

    def reason(n):
        n = n.lower()
        tags = [lab for pat, lab in (("frameshift", "frameshift"), ("internal stop", "internal stop"),
                                     ("incomplete|partial|truncat|missing", "incomplete/truncated"),
                                     ("premature", "premature stop")) if re.search(pat, n)]
        return "; ".join(tags) if tags else "unspecified"

    p["reason"] = p["note_all"].map(reason)
    save_table(p[["seqid", "start", "end", "strand", "length", "locus_tag", "product", "reason", "note_all"]],
               "04_pseudogenes.csv")
    say(f"\nPseudogenes: {len(p)}")
    if len(p):
        say("  by reason: " + str(p["reason"].value_counts().to_dict()))
        say(f"  median pseudogene length {p['length'].median():.0f} bp vs "
            f"{genes[genes['class'] == 'protein_coding']['length'].median():.0f} bp for protein-coding genes")

    # ---- shared protein ids
    cds = gff[(gff["type"] == "CDS") & (gff["protein_id"] != "") & (gff["pseudo"].str.lower() != "true")]
    cds = cds.drop_duplicates("ID")
    dup = cds[cds.duplicated("protein_id", keep=False)].sort_values(["protein_id", "start"])
    save_table(dup[["protein_id", "seqid", "start", "end", "strand", "locus_tag", "product"]], "04_shared_protein_ids.csv")
    say(f"\nProtein IDs used by more than one CDS: {dup['protein_id'].nunique()} IDs "
        f"({len(dup)} CDS features; {len(dup) - dup['protein_id'].nunique()} extra copies)")
    if len(dup):
        show(dup[["protein_id", "seqid", "start", "end", "strand", "product"]], 10)
    return t


# ----------------------------------------------------------------------
# 5. Codon usage
# ----------------------------------------------------------------------
BASES = "TCAG"
AA_STR = "FFLLSSSSYY**CC*WLLLLPPPPHHQQRRRRIIIMTTTTNNKKSSRRVVVVAAAADDEEGGGG"
CODONS = ["".join(c) for c in itertools.product(BASES, repeat=3)]
CIDX = {c: i for i, c in enumerate(CODONS)}
CODON_AA = dict(zip(CODONS, AA_STR))
FAM = defaultdict(list)
for _c in CODONS:
    if CODON_AA[_c] != "*":
        FAM[CODON_AA[_c]].append(CIDX[_c])
CLASSES = {k: [a for a in FAM if len(FAM[a]) == k] for k in (2, 3, 4, 6)}
SENSE_MASK = np.array([CODON_AA[c] not in "*MW" for c in CODONS])
GC3_MASK = np.array([c[2] in "GC" for c in CODONS])


def extract_cds(gff, genome):
    cds = gff[gff["type"] == "CDS"].copy()
    cds["gid"] = np.where(cds["ID"] != "", cds["ID"], cds["Parent"])
    seqs, meta, skipped = [], [], Counter()
    for gid, d in cds.groupby("gid", sort=False):
        f = d.iloc[0]
        if f["pseudo"].lower() == "true":
            skipped["pseudogene"] += 1
            continue
        if "true" in f["partial"].lower():
            skipped["partial"] += 1
            continue
        if f["seqid"] not in genome:
            skipped["sequence not in fna"] += 1
            continue
        d = d.sort_values("start")
        s = "".join(genome[f["seqid"]][a - 1:b] for a, b in zip(d["start"], d["end"]))
        if f["strand"] == "-":
            s = revcomp(s)
        phase = d.iloc[0]["phase"] if f["strand"] == "+" else d.iloc[-1]["phase"]
        if phase in ("1", "2"):
            s = s[int(phase):]
        if len(s) % 3 or len(s) < 30:
            skipped["bad length"] += 1
            continue
        if s[-3:] in ("TAA", "TAG", "TGA"):
            s = s[:-3]
        seqs.append(s)
        meta.append({"gid": gid, "seqid": f["seqid"], "start": d["start"].min(), "end": d["end"].max(),
                     "strand": f["strand"], "locus_tag": f["locus_tag"], "product": f["product"]})
    return seqs, pd.DataFrame(meta), skipped


def codon_matrix(seqs):
    M = np.zeros((len(seqs), 64))
    for i, s in enumerate(seqs):
        for k in range(0, len(s) - 2, 3):
            j = CIDX.get(s[k:k + 3])
            if j is not None:
                M[i, j] += 1
    return M


def rscu(v):
    out = np.zeros(64)
    for aa, idx in FAM.items():
        tot = v[idx].sum()
        if tot > 0:
            out[idx] = v[idx] * len(idx) / tot
    return out


def enc_row(row):
    Fk = {}
    for k, aas in CLASSES.items():
        num = den = 0.0
        for a in aas:
            cnt = row[FAM[a]]
            n = cnt.sum()
            if n > 1:
                F = (n * np.sum((cnt / n) ** 2) - 1) / (n - 1)
                num += n * F
                den += n
        Fk[k] = num / den if den else np.nan
    if np.isnan(Fk[3]):
        Fk[3] = np.nanmean([Fk[2], Fk[4]])
    if any(np.isnan(v) for v in Fk.values()):
        return np.nan
    Fk = {k: max(v, 1e-3) for k, v in Fk.items()}
    return min(2 + 9 / Fk[2] + 1 / Fk[3] + 5 / Fk[4] + 3 / Fk[6], 61.0)


def codon_analysis(gff, genome, trna):
    heading("5. CODON USAGE")
    seqs, meta, skipped = extract_cds(gff, genome)
    say(f"CDS used for codon analysis: {len(seqs):,}; skipped: {dict(skipped)}")
    if len(seqs) < 20:
        say("Too few CDS for codon analysis; skipped.")
        return
    M = codon_matrix(seqs)
    meta["codons"] = M.sum(axis=1)
    meta["GC_%"] = [round(gc_frac(s) * 100, 2) for s in seqs]
    sense = M[:, SENSE_MASK].sum(axis=1)
    meta["GC3s_%"] = np.where(sense > 0, M[:, SENSE_MASK & GC3_MASK].sum(axis=1) / np.maximum(sense, 1) * 100, np.nan)
    ok = (meta["codons"] >= MIN_CODONS_PER_GENE).values
    meta["ENC"] = [enc_row(M[i]) if ok[i] else np.nan for i in range(len(seqs))]

    ribo = meta["product"].str.contains(r"ribosomal protein", case=False, regex=True).values.copy()
    ribo &= ok
    tot_rscu = rscu(M.sum(axis=0))
    table = pd.DataFrame({"codon": CODONS, "aa": [CODON_AA[c] for c in CODONS],
                          "count": M.sum(axis=0).astype(int), "RSCU_all": tot_rscu.round(3)})
    table["fraction_%"] = (table["count"] / table["count"].sum() * 100).round(3)

    if ribo.sum() >= 10:
        r_rscu = rscu(M[ribo].sum(axis=0))
        table["RSCU_ribosomal"] = r_rscu.round(3)
        w = np.ones(64)
        for aa, idx in FAM.items():
            mx = r_rscu[idx].max()
            if mx > 0:
                w[idx] = np.maximum(r_rscu[idx] / mx, 0.01)
        lw = np.log(w) * SENSE_MASK
        meta["CAI"] = np.exp((M * lw).sum(axis=1) / np.maximum((M[:, SENSE_MASK]).sum(axis=1), 1))
    else:
        meta["CAI"] = np.nan
        say("  <10 ribosomal-protein genes found; CAI skipped.")
    save_table(table, "05_codon_usage_table.csv")

    z = (meta["GC3s_%"] - meta["GC3s_%"].mean()) / meta["GC3s_%"].std()
    meta["GC3s_zscore"] = z.round(2)
    meta["atypical_GC3"] = z.abs() > 2
    meta["ribosomal"] = ribo
    save_table(meta.drop(columns=["gid"]), "05_gene_codon_metrics.csv")
    save_table(meta[meta["atypical_GC3"]].sort_values("GC3s_zscore", key=abs, ascending=False)
               .drop(columns=["gid"]), "05_atypical_genes.csv")

    say(f"Mean GC of CDS: {meta['GC_%'].mean():.1f}%; mean GC3s: {meta['GC3s_%'].mean():.1f}%")
    say(f"Mean ENC: {meta['ENC'].mean():.1f}" + (f"; ribosomal genes ENC {meta.loc[ribo, 'ENC'].mean():.1f}" if ribo.any() else ""))
    say(f"Atypical genes (|GC3s z| > 2): {int(meta['atypical_GC3'].sum())} "
        f"({meta['atypical_GC3'].mean() * 100:.1f}%) - candidates for horizontally transferred DNA")

    fig, ax = plt.subplots(1, 3, figsize=(17, 4.5))
    ax[0].scatter(meta["GC3s_%"], meta["GC_%"], s=4, c=np.where(meta["atypical_GC3"], "red", "steelblue"), alpha=0.6)
    ax[0].set_xlabel("GC3s (%)")
    ax[0].set_ylabel("Gene GC (%)")
    ax[0].set_title("GC vs GC3s (red = atypical)")
    s = np.linspace(0.01, 0.99, 200)
    ax[1].plot(s * 100, 2 + s + 29 / (s ** 2 + (1 - s) ** 2), "k--", lw=1, label="Expected (no selection)")
    ax[1].scatter(meta["GC3s_%"], meta["ENC"], s=4, alpha=0.5, label="genes")
    if ribo.any():
        ax[1].scatter(meta.loc[ribo, "GC3s_%"], meta.loc[ribo, "ENC"], s=14, color="red", label="ribosomal")
    ax[1].set_xlabel("GC3s (%)")
    ax[1].set_ylabel("ENC")
    ax[1].set_title("Wright plot")
    ax[1].legend(fontsize=8)
    if meta["CAI"].notna().any():
        ax[2].hist(meta["CAI"].dropna(), bins=50, color="grey")
        if ribo.any():
            ax[2].hist(meta.loc[ribo, "CAI"], bins=20, color="red", alpha=0.7, label="ribosomal")
            ax[2].legend()
        ax[2].set_title("CAI (reference = ribosomal genes)")
        ax[2].set_xlabel("CAI")
    save_fig(fig, "05_codon_usage.png")

    # tRNA supply vs codon demand
    if trna is not None and len(trna):
        aa_tot = pd.Series({aa: M.sum(axis=0)[idx].sum() for aa, idx in FAM.items()})
        demand = (aa_tot / aa_tot.sum() * 100).rename("codon_demand_%")
        supply = trna.dropna(subset=["aa1"]).groupby("aa1").size().rename("tRNA_genes")
        anti = trna.dropna(subset=["aa1"]).groupby("aa1")["anticodon_seq"].agg(
            lambda x: ",".join(sorted(set(v for v in x if v)))).rename("anticodons")
        comp = pd.concat([demand, supply, anti], axis=1).fillna({"tRNA_genes": 0, "anticodons": ""})
        comp["tRNA_genes"] = comp["tRNA_genes"].astype(int)
        comp = comp.loc[[a for a in "ACDEFGHIKLMNPQRSTVWY" if a in comp.index]]
        save_table(comp.reset_index().rename(columns={"index": "aa"}), "05_trna_vs_codon_demand.csv")
        rx, ry = comp["codon_demand_%"].rank(), comp["tRNA_genes"].rank()
        rho = np.corrcoef(rx, ry)[0, 1]
        say(f"\nSpearman correlation, tRNA gene copies vs amino-acid demand: {rho:.2f}")
        fig, ax = plt.subplots(1, 2, figsize=(13, 4.5))
        ax[0].bar(comp.index, comp["tRNA_genes"], color="seagreen")
        ax[0].set_title("tRNA gene copies per amino acid")
        ax[1].scatter(comp["codon_demand_%"], comp["tRNA_genes"], color="seagreen")
        for aa, row in comp.iterrows():
            ax[1].annotate(aa, (row["codon_demand_%"], row["tRNA_genes"]), fontsize=8, xytext=(3, 3),
                           textcoords="offset points")
        ax[1].set_xlabel("Amino-acid usage in CDS (%)")
        ax[1].set_ylabel("tRNA gene copies")
        ax[1].set_title(f"tRNA supply vs demand (Spearman r = {rho:.2f})")
        save_fig(fig, "05_trna_vs_demand.png")


# ----------------------------------------------------------------------
# 6. Proteins
# ----------------------------------------------------------------------
RES_MASS = {"A": 71.0788, "R": 156.1875, "N": 114.1038, "D": 115.0886, "C": 103.1388, "E": 129.1155,
            "Q": 128.1307, "G": 57.0519, "H": 137.1411, "I": 113.1594, "L": 113.1594, "K": 128.1741,
            "M": 131.1926, "F": 147.1766, "P": 97.1167, "S": 87.0782, "T": 101.1051, "W": 186.2132,
            "Y": 163.1760, "V": 99.1326}
KD = {"A": 1.8, "R": -4.5, "N": -3.5, "D": -3.5, "C": 2.5, "Q": -3.5, "E": -3.5, "G": -0.4, "H": -3.2,
      "I": 4.5, "L": 3.8, "K": -3.9, "M": 1.9, "F": 2.8, "P": -1.6, "S": -0.8, "T": -0.7, "W": -0.9,
      "Y": -1.3, "V": 4.2}
POS = {"K": 10.8, "R": 12.5, "H": 6.5}
NEG = {"D": 3.9, "E": 4.1, "C": 8.5, "Y": 10.1}


def isoelectric_point(cnt):
    def charge(ph):
        pos = 1 / (1 + 10 ** (ph - 8.6)) + sum(cnt.get(a, 0) / (1 + 10 ** (ph - pk)) for a, pk in POS.items())
        neg = 1 / (1 + 10 ** (3.6 - ph)) + sum(cnt.get(a, 0) / (1 + 10 ** (pk - ph)) for a, pk in NEG.items())
        return pos - neg
    lo, hi = 0.0, 14.0
    for _ in range(35):
        mid = (lo + hi) / 2
        lo, hi = (mid, hi) if charge(mid) > 0 else (lo, mid)
    return (lo + hi) / 2


def protein_analysis(faa_path):
    heading("6. PROTEINS (FAA)")
    prot, headers = read_fasta(faa_path)
    rows, comp = [], Counter()
    for pid, seq in prot.items():
        seq = seq.replace("*", "").upper()
        valid = [a for a in seq if a in RES_MASS]
        if not valid:
            continue
        cnt = Counter(valid)
        comp.update(cnt)
        n = len(valid)
        prod = re.sub(r"\s*\[.*?\]\s*$", "", headers[pid].split(" ", 1)[1] if " " in headers[pid] else "")
        rows.append({"protein_id": pid, "product": prod, "length_aa": len(seq),
                     "MW_kDa": round((sum(RES_MASS[a] * c for a, c in cnt.items()) + 18.015) / 1000, 2),
                     "pI": round(isoelectric_point(cnt), 2),
                     "GRAVY": round(sum(KD[a] * c for a, c in cnt.items()) / n, 3)})
    df = pd.DataFrame(rows)
    df["hypothetical"] = df["product"].str.contains("hypothetical", case=False)
    save_table(df, "06_protein_properties.csv")
    say(f"Proteins: {len(df):,}; median length {df['length_aa'].median():.0f} aa; "
        f"mean MW {df['MW_kDa'].mean():.1f} kDa; median pI {df['pI'].median():.2f}")
    say(f"Hypothetical / uncharacterised proteins: {int(df['hypothetical'].sum()):,} "
        f"({df['hypothetical'].mean() * 100:.1f}%)")
    show(df.sort_values("length_aa", ascending=False)[["protein_id", "product", "length_aa", "MW_kDa", "pI"]], 5)

    tot = sum(comp.values())
    aa = pd.Series({a: comp[a] / tot * 100 for a in "ACDEFGHIKLMNPQRSTVWY"})
    save_table(aa.rename("percent").reset_index().rename(columns={"index": "aa"}), "06_aa_composition.csv")

    fig, ax = plt.subplots(2, 3, figsize=(16, 8))
    ax = ax.ravel()
    ax[0].hist(df["length_aa"].clip(upper=1500), bins=60, color="steelblue")
    ax[0].set_title(f"Protein length (median {df['length_aa'].median():.0f} aa)")
    ax[1].hist(df["MW_kDa"].clip(upper=150), bins=60, color="slateblue")
    ax[1].set_title("Molecular weight (kDa)")
    ax[2].hist(df["pI"], bins=60, color="darkorange")
    ax[2].set_title("Isoelectric point (pI)")
    ax[3].hist(df["GRAVY"], bins=60, color="seagreen")
    ax[3].set_title("Hydropathy (GRAVY)")
    ax[4].bar(aa.index, aa.values, color="grey")
    ax[4].set_title("Amino-acid composition (%)")
    ax[5].pie([int((~df["hypothetical"]).sum()), int(df["hypothetical"].sum())],
              labels=["Annotated", "Hypothetical"], autopct="%1.1f%%", colors=["#4c72b0", "#c44e52"])
    ax[5].set_title("Functional annotation status")
    save_fig(fig, "06_protein_properties.png")
    return df


# ----------------------------------------------------------------------
# 7. Circular genome map
# ----------------------------------------------------------------------
def circular_map(sid, seq, genes, ori_ter, organism, fname):
    L = len(seq)
    pos, gc, skew = window_stats(seq, max(auto_window(L) // 2, 200))
    tw = 2 * np.pi / L
    fig = plt.figure(figsize=(9, 9))
    ax = fig.add_subplot(111, projection="polar")
    ax.set_theta_zero_location("N")
    ax.set_theta_direction(-1)

    def ring(d, r0, r1, color):
        if len(d):
            width = np.maximum((d["end"].values - d["start"].values + 1) * tw, 2 * np.pi / 2500)
            ax.bar(d["start"].values * tw, r1 - r0, width=width, bottom=r0, align="edge",
                   color=color, linewidth=0)

    g = genes[genes["seqid"] == sid]
    pc = g[~g["class"].isin(["pseudogene", "rRNA", "tRNA"])]
    ring(pc[pc["strand"] == "+"], 8.6, 9.5, "#3b6fb6")
    ring(pc[pc["strand"] == "-"], 7.7, 8.6, "#d1495b")
    ring(g[g["class"] == "pseudogene"], 7.7, 9.5, "#f4a259")
    ring(g[g["class"] == "tRNA"], 9.6, 10.3, "#2a9d8f")
    ring(g[g["class"] == "rRNA"], 9.6, 10.6, "black")

    for base, vals, label in ((6.4, (gc - np.nanmean(gc)), "GC"), (4.8, np.nan_to_num(skew), "skew")):
        amp = np.nanmax(np.abs(vals)) or 1
        r = base + vals / amp * 0.9
        th = pos * tw
        ax.plot(np.linspace(0, 2 * np.pi, 300), np.full(300, base), color="grey", lw=0.4)
        ax.fill_between(th, base, r, where=vals >= 0, color="#1b9e77", alpha=0.8, interpolate=True)
        ax.fill_between(th, base, r, where=vals < 0, color="#7570b3", alpha=0.8, interpolate=True)

    if sid in ori_ter:
        for p, lab, col in ((ori_ter[sid][0], "oriC", "red"), (ori_ter[sid][1], "ter", "blue")):
            ax.plot([p * tw] * 2, [3.8, 10.8], color=col, ls="--", lw=1)
            ax.text(p * tw, 11.2, lab, color=col, ha="center", fontsize=9)

    step = 10 ** int(math.floor(math.log10(L / 6)))
    step = step * (5 if L / step > 12 else 2 if L / step > 8 else 1) if step else 1
    ticks = np.arange(step, L, step)   # starts at 'step' so no "0.00 Mb" label is drawn
    ax.set_xticks(ticks * tw)
    ax.set_xticklabels([f"{t / 1e6:.2f} Mb" if L > 1e6 else f"{t / 1e3:.0f} kb" for t in ticks], fontsize=8)
    ax.set_yticks([])
    ax.set_ylim(0, 11.5)
    ax.grid(False)
    ax.spines["polar"].set_visible(False)
    ax.text(0, 0, f"{organism}\n{sid}\n{L:,} bp\nGC {gc_frac(seq) * 100:.1f}%\n{len(g):,} genes",
            ha="center", va="center", fontsize=10)
    from matplotlib.patches import Patch
    ax.legend(handles=[Patch(color="#3b6fb6", label="CDS +"), Patch(color="#d1495b", label="CDS -"),
                       Patch(color="black", label="rRNA"), Patch(color="#2a9d8f", label="tRNA"),
                       Patch(color="#f4a259", label="pseudogene"),
                       Patch(color="#1b9e77", label="GC / skew above mean"),
                       Patch(color="#7570b3", label="below mean")],
              loc="lower left", bbox_to_anchor=(-0.12, -0.08), fontsize=8, frameon=False)
    ax.set_title("Circular genome map (outer to inner: RNA, CDS +, CDS -, GC content, GC skew)", fontsize=10, pad=20)
    save_fig(fig, fname)


# ----------------------------------------------------------------------
# Main
# ----------------------------------------------------------------------
def main():
    OUT.mkdir(exist_ok=True)
    found = get_input_files()

    # ---- load
    meta, rep_tbl = ({}, None)
    if "report" in found:
        meta, rep_tbl = parse_assembly_report(found["report"])
    genome, _ = read_fasta(found["fna"])
    genome = {k: v.upper() for k, v in genome.items()}
    gff = parse_gff(found["gff"])
    genes = build_gene_table(gff)

    # ---- 1. overview
    heading("1. ASSEMBLY OVERVIEW")
    organism = meta.get("Organism name", "").split(" (")[0] or "Genome"
    for key in ("Organism name", "Infraspecific name", "Assembly name", "Assembly level",
                "Genome representation", "RefSeq assembly accession", "Assembly method"):
        if key in meta:
            say(f"{key}: {meta[key]}")
    rep = replicon_table(genome, rep_tbl)
    save_table(rep, "01_replicons.csv")
    show(rep, 20)
    total = rep["length_bp"].sum()
    say(f"\nTotal length: {total:,} bp ({total / 1e6:.2f} Mb); replicons/contigs: {len(rep)}; "
        f"N50: {n50(rep['length_bp']):,} bp")
    say(f"Overall GC: {sum(genome[s].count('G') + genome[s].count('C') for s in genome) / sum(len(v) for v in genome.values()) * 100:.2f}%"
        f"; ambiguous bases: {int(rep['N_count'].sum() + rep['other_ambiguous'].sum()):,}")
    missing = set(gff["seqid"]) - set(genome)
    if missing:
        say(f"WARNING: GFF refers to sequence IDs not found in the .fna: {sorted(missing)[:5]}")
    plot_overview(rep)

    # ---- 2-5
    ori_ter, _ = composition_analysis(genome, rep)
    gene_structure(genes, gff, rep, ori_ter)
    trna = rna_analysis(gff, genes, rep, ori_ter)
    codon_analysis(gff, genome, trna)

    # ---- 6
    if "faa" in found:
        protein_analysis(found["faa"])

    # ---- 7
    heading("7. CIRCULAR GENOME MAPS")
    todo = rep[rep["length_bp"] >= MIN_REPLICON_FOR_MAP].head(MAX_MAPS)
    for i, (_, r) in enumerate(todo.iterrows(), 1):
        circular_map(r["seqid"], genome[r["seqid"]], genes, ori_ter, organism, f"07_circular_map_{i}.png")
    if todo.empty:
        say("No replicon >= MIN_REPLICON_FOR_MAP; no circular map drawn.")

    # ---- wrap up
    (OUT / "summary.txt").write_text("\n".join(SUMMARY))
    heading("DONE")
    say(f"Tables, figures and summary.txt are in the '{OUT}' folder.")
    try:
        from google.colab import files
        shutil.make_archive("results", "zip", OUT)
        files.download("results.zip")
    except ImportError:
        pass


main()
```
